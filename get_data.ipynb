```python
users = df[df["role"].str.lower() == "user"].copy()
```

```python
users["message"] = users["message"].fillna("").astype(str)
```

```python
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
```

```python
phone_pattern = r'\d{10}'
```

```python
users = users[
    ~users["message"].str.contains(email_pattern, regex=True, na=False) &
    ~users["message"].str.contains(phone_pattern, regex=True, na=False)
]
```

```python
with open("user_messages.txt", "w", encoding="utf-8") as f:
    for message in users["message"]:
        f.write(message.strip() + "\n")

print(f"Saved {len(users)} messages.")
```
